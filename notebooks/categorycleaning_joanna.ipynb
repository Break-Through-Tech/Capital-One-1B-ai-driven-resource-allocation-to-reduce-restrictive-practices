### This file holds the code for the cleaning of the `School Support` and `Referrals and Arrests` school level data across 2017-2024 (9th-12th grade)

First we need to find what columns exist in the files across all the years:


```python
%pip install pandas
import glob
import pandas as pd

files = glob.glob("../data/filtered_data/school_support/*.csv")
print(len(files))
overlap_columns = {}

individual_columns = {}
# there are only 4 files/ so the columns should exist across all four

for f in files:
    df = pd.read_csv(f)
    col_list = df.columns
    individual_columns[f] = col_list
    for i in col_list:
        if i not in overlap_columns:
            overlap_columns[i] = 1
        else:
            overlap_columns[i] += 1

true_overlap_columns_support = {k: v for k, v in overlap_columns.items() if v == len(files)}

print(true_overlap_columns_support)

# print(individual_columns)
```

#### Replacing negative values with NaN

```python
# ------
# change negative values and treat them as missing data
# ------
import os
import numpy as np

folder = "../data/filtered_data/school_support_overlap/"
out_folder = "../data/cleaned/school_support/"
os.makedirs(out_folder, exist_ok=True)

files = glob.glob(folder + "*.csv")

for f in files:
    df = pd.read_csv(f, dtype={"LEAID": str, "SCHID": str, "COMBOKEY": str,
                               "LEA_NAME": str, "SCH_NAME": str,
                               "LEA_STATE_NAME": str, "LEA_STATE": str})

    # go through the count columns one at a time
    for c in data_cols:
        negative_rows = df[c] < 0          # True where the value is negative
        df.loc[negative_rows, c] = np.nan  # in those rows, set this column to missing

    # check: no negatives should be left
    leftover = 0
    for c in data_cols:
        leftover = leftover + (df[c] < 0).sum()
    print(f.split("/")[-1], "negatives left:", leftover)

    df.to_csv(out_folder + f.split("/")[-1], index=False)
```

### Dealing with inconsistent data

```python
# Checking data after replacing missing values
cleaned_files_support = glob.glob("../data/cleaned/school_support/*.csv")

# dropping mismatched rows where certified + non-certified does not equal what was reported
# in the total column
for f in cleaned_files_support:
    df = pd.read_csv(f)
    yr = df["YEAR"].iloc[0]
    
    valid_rows = df['SCH_FTETEACH_CERT'].notna() & df['SCH_FTETEACH_NOTCERT'].notna() & df['SCH_FTETEACH_TOT'].notna()
    subset = df.loc[valid_rows].copy()
    
    # Check where certified + non-certified does not equal the total column
    mismatches = subset[abs((subset['SCH_FTETEACH_CERT'] + subset['SCH_FTETEACH_NOTCERT']) - subset['SCH_FTETEACH_TOT']) > 0.01]
    df.drop(mismatches.index, inplace=True)
    df.to_csv(f, index=False)

# checking the sums after dropping the mismatched rows
for f in cleaned_files_support:
    df = pd.read_csv(f)
    yr = df["YEAR"].iloc[0]
    
    # Filter for rows where all 3 relevant columns are valid numbers
    valid_rows = df['SCH_FTETEACH_CERT'].notna() & df['SCH_FTETEACH_NOTCERT'].notna() & df['SCH_FTETEACH_TOT'].notna()
    
    cert = df.loc[valid_rows, "SCH_FTETEACH_CERT"].sum()
    noncert = df.loc[valid_rows, "SCH_FTETEACH_NOTCERT"].sum()
    tot = df.loc[valid_rows, "SCH_FTETEACH_TOT"].sum()

    print(f"YEAR: {yr}, CERT: {cert}, NONCERT: {noncert}, TOT: {tot}, SUM (Cert+Noncert): {cert + noncert}")
```

```python

cleaned_files_support = glob.glob("../data/cleaned/school_support/*.csv")

for f in cleaned_files_support:
    df = pd.read_csv(f)
    print(f"\n--- Year: {df['YEAR'].iloc[0]} ---")

    sorted_df = pd.DataFrame()
    data_cols = [c for c in numeric_cols if c != "YEAR" and c != "SCHID"]
    for col in data_cols:
        if col in df.columns:
        
            clean_vals = df[col].dropna().sort_values(ascending=True).reset_index(drop=True)
            sorted_df[col] = clean_vals
            
    # Get the median for every numeric column
    medians = df.select_dtypes(include='number').median()
    print(f"Medians for {df['YEAR'].iloc[0]}: {medians}")
    # Calculate Q1 and Q3 for all numeric columns
    quartiles = df.select_dtypes(include='number').quantile([0.25, 0.75])
    print(f"Quartiles for {df['YEAR'].iloc[0]}: {quartiles}")

    print(f"Interquartile Range (IQR) for {df['YEAR'].iloc[0]}:")
    for col in data_cols:
        q1 = quartiles.loc[0.25, col]
        q3 = quartiles.loc[0.75, col]  
        iqr = q3 - q1
        print(f"  {col}: {iqr}")

        upper_fence = q3 + (1.5 * iqr)
        lower_fence = q1 - (1.5 * iqr)

        print(f"  {col} - Upper Fence: {upper_fence}, Lower Fence: {lower_fence}")
```

```python
import pandas as pd
import glob

cleaned_files_support = glob.glob("../data/cleaned/school_support/*.csv")

for f in cleaned_files_support:
    df = pd.read_csv(f)
    yr = df["YEAR"].iloc[0] if "YEAR" in df.columns and not pd.isna(df["YEAR"].iloc[0]) else "Unknown Year"
    print(f"\n==============================")
    print(f"Inspection for Year: {yr}")
    print(f"==============================")
    
    col = 'SCH_FTETEACH_TOT'
    if col in df.columns:
        # Calculate Q1, Q3, and Upper Fence
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1
        upper_fence = q3 + (1.5 * iqr)
        
        # Filter schools exceeding the upper fence
        outliers = df[df[col] > upper_fence]
        
        print(f"Upper Fence Threshold: {upper_fence:.2f}")
        print(f"Total schools exceeding upper fence: {len(outliers)}")
        print("\nTop 10 highest-staffed schools above the fence:")
        
        # Sort by highest values first and display key identifying columns
        display_cols = [c for c in ['SCH_NAME', 'LEA_NAME', 'LEA_STATE', col] if c in df.columns]
        top_outliers = outliers.sort_values(by=col, ascending=False)
        
        print(top_outliers[display_cols].head(10).to_string(index=False))
```
