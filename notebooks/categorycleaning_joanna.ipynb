#### Replacing negative values with NaN

```python
# ------
# change negative values and treat them as missing data
# ------
import os
import numpy as np

folder = "../data/filtered_data/school_support_overlap/"
out_folder = "../data/cleaned/school_support/"
os.makedirs(out_folder, exist_ok=True)

files = glob.glob(folder + "*.csv")

for f in files:
    df = pd.read_csv(f, dtype={"LEAID": str, "SCHID": str, "COMBOKEY": str,
                               "LEA_NAME": str, "SCH_NAME": str,
                               "LEA_STATE_NAME": str, "LEA_STATE": str})
    numeric_cols = df.select_dtypes("number").columns
    data_cols = [c for c in numeric_cols if c != "YEAR"]
    # go through the count columns one at a time
    for c in data_cols:
        negative_rows = df[c] < 0          # True where the value is negative
        df.loc[negative_rows, c] = np.nan  # in those rows, set this column to missing

    # check: no negatives should be left
    leftover = 0
    for c in data_cols:
        leftover = leftover + (df[c] < 0).sum()
    print(f.split("/")[-1], "negatives left:", leftover)

    df.to_csv(out_folder + f.split("/")[-1], index=False)
```

### Dealing with inconsistent data

```python
# Checking data after replacing missing values
cleaned_files_support = glob.glob("../data/cleaned/school_support/*.csv")

# dropping mismatched rows where certified + non-certified does not equal what was reported
# in the total column
for f in cleaned_files_support:
    df = pd.read_csv(f)
    yr = df["YEAR"].iloc[0]
    
    valid_rows = df['SCH_FTETEACH_CERT'].notna() & df['SCH_FTETEACH_NOTCERT'].notna() & df['SCH_FTETEACH_TOT'].notna()
    subset = df.loc[valid_rows].copy()
    
    # Check where certified + non-certified does not equal the total column
    mismatches = subset[abs((subset['SCH_FTETEACH_CERT'] + subset['SCH_FTETEACH_NOTCERT']) - subset['SCH_FTETEACH_TOT']) > 0.01]
    df.drop(mismatches.index, inplace=True)
    df.to_csv(f, index=False)

# checking the sums after dropping the mismatched rows
for f in cleaned_files_support:
    df = pd.read_csv(f)
    yr = df["YEAR"].iloc[0]
    
    # Filter for rows where all 3 relevant columns are valid numbers
    valid_rows = df['SCH_FTETEACH_CERT'].notna() & df['SCH_FTETEACH_NOTCERT'].notna() & df['SCH_FTETEACH_TOT'].notna()
    
    cert = df.loc[valid_rows, "SCH_FTETEACH_CERT"].sum()
    noncert = df.loc[valid_rows, "SCH_FTETEACH_NOTCERT"].sum()
    tot = df.loc[valid_rows, "SCH_FTETEACH_TOT"].sum()

    print(f"YEAR: {yr}, CERT: {cert}, NONCERT: {noncert}, TOT: {tot}, SUM (Cert+Noncert): {cert + noncert}")
```

```python
import pandas as pd
import numpy as np
import glob

# Dictionary mapping year to enrollment file path
enrollment_files = {
    "2017-2018": "../notebooks/extra_data/Enrollment_1718.csv",
    "2020-2021": "../notebooks/extra_data/Enrollment_2021.csv",
    "2021-2022": "../notebooks/extra_data/Enrollment_2122.csv",
    "2023-2024": "../notebooks/extra_data/Enrollment_2324.csv"
}

referral_files = glob.glob("../data/cleaned/referrals_arrests/*.csv")

for rf in referral_files:
    df_ref = pd.read_csv(rf, dtype={"COMBOKEY": str, "SCHID": str, "LEAID": str})
    yr = str(df_ref["YEAR"].iloc[0]) if "YEAR" in df_ref.columns else None
    
    print(f"\n==========================================")
    print(f"Processing Year: {yr}")
    print(f"==========================================")
    
    if yr in enrollment_files:
        df_enroll = pd.read_csv(enrollment_files[yr], dtype={"COMBOKEY": str, "SCHID": str, "LEAID": str})
        
        # 1. Clean negative values in enrollment sub-total columns
        sub_tot_cols = [c for c in ['TOT_ENR_M', 'TOT_ENR_F', 'TOT_ENR_X'] if c in df_enroll.columns]
        for c in sub_tot_cols:
            df_enroll[c] = df_enroll[c].apply(lambda x: np.nan if x < 0 else x)
            
        # 2. Compute unified TOT_ENR if not already present
        if 'TOT_ENR' not in df_enroll.columns:
            df_enroll['TOT_ENR'] = df_enroll[sub_tot_cols].sum(axis=1, skipna=True)
        else:
            df_enroll['TOT_ENR'] = df_enroll['TOT_ENR'].apply(lambda x: np.nan if x < 0 else x)
            
        # 3. Merge TOT_ENR into referrals dataframe on COMBOKEY
        df_merged = df_ref.merge(
            df_enroll[['COMBOKEY', 'TOT_ENR']], 
            on='COMBOKEY', 
            how='left'
        )
        
        # 4. Calculate rate per 100 students (e.g., for Total Male Disciplinary Referrals)
        target_cols = ['TOT_DISCWODIS_REF_M', 'TOT_DISCWODIS_REF_F', 
                       'TOT_DISCWODIS_ARR_M', 'TOT_DISCWODIS_ARR_F']
        
        for col in target_cols:
            if col in df_merged.columns:
                rate_col = f"{col}_PER_100"
                # Avoid division by 0
                df_merged[rate_col] = (df_merged[col] / df_merged['TOT_ENR'].replace(0, np.nan)) * 100
                
                # 1. Isolate schools with active (non-zero) rates
                active_schools = df_merged[df_merged[rate_col] > 0][rate_col]
                
                if len(active_schools) > 0:
                    # 2. Compute Q1 and Q3 ONLY on active schools
                    q1 = active_schools.quantile(0.25)
                    q3 = active_schools.quantile(0.75)
                    iqr = q3 - q1
                    upper_fence = q3 + (3.0 * iqr)
                    
                    # 3. Find outliers exceeding the fence among all schools
                    outliers = df_merged[df_merged[rate_col] > upper_fence]
                
                    print(f"Metric: {rate_col} | Upper Fence: {upper_fence:.2f}% | Extreme Outliers: {len(outliers)}")
    else:
        print(f"Warning: Enrollment file not found for year {yr}")
```

```python
import glob
import pandas as pd

folder = "../data/cleaned/school_support"
paths = glob.glob(f"{folder}/*.csv")

frames = []
for path in paths:
    df = pd.read_csv(path, dtype={"COMBOKEY": str, "LEAID": str, "SCHID": str})
    frames.append(df)

stacked = pd.concat(frames, ignore_index=True)

# COMBOKEY must appear across all 4 years ( so that we can assess the trend of risk)
n_years = stacked.groupby("COMBOKEY")["YEAR"].nunique()
keep_ids = n_years[n_years == stacked["YEAR"].nunique()].index

merged = stacked[stacked["COMBOKEY"].isin(keep_ids)].copy()

print(merged["YEAR"].value_counts())
print("unique schools:", merged["COMBOKEY"].nunique())
merged.to_csv("../data/cleaned/merged_schoolsupport.csv", index=False)
```
